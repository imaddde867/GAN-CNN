```python
import os 
import pandas as pd

# defining the data directories
train_image_dir = 'data/train/images'
train_label_dir = 'data/train/labels'
valid_image_dir = 'data/valid/images'
valid_label_dir = 'data/valid/labels'

# loading the data
def load_files():
    """Load all files in a dir"""
    train_images = [f for f in os.listdir(train_image_dir)]
    train_labels = [f for f in os.listdir(train_label_dir)]
    valid_images = [f for f in os.listdir(valid_image_dir)]
    valid_labels = [f for f in os.listdir(valid_label_dir)]
    
    return train_images, train_labels, valid_images, valid_labels

train_images, train_labels, valid_images, valid_labels = load_files()

print(f"Loaded: {len(train_images)} training images")
print(f"Loaded: {len(train_labels)} training labels")
print(f"Loaded: {len(valid_images)} validation images")
print(f"Loaded: {len(valid_labels)} validation labels")
```

```python
# Create DataFrame with dataset info
df = pd.DataFrame({
    'filename': train_images + train_labels + valid_images + valid_labels,
    'dataset': (['train'] * len(train_images)) + 
               (['train'] * len(train_labels)) + 
               (['valid'] * len(valid_images)) + 
               (['valid'] * len(valid_labels)),
    'type': (['image'] * len(train_images)) + 
            (['label'] * len(train_labels)) + 
            (['image'] * len(valid_images)) + 
            (['label'] * len(valid_labels))
})

# Extract filename without extension to create matching IDs
df['id'] = df['filename'].apply(lambda x: os.path.splitext(x)[0])

# Extract extension
df['extension'] = df['filename'].apply(lambda x: os.path.splitext(x)[1])

print("\nDataFrame preview:")
print(df.head(10))
```

```python
# Check for matching image-label pairs
print("\nChecking for matches : ")
image_ids = set(df[df['type'] == 'image']['id'])
label_ids = set(df[df['type'] == 'label']['id'])

matched = image_ids & label_ids
unmatched_images = image_ids - label_ids
unmatched_labels = label_ids - image_ids

print(f"\nMatched pairs: {len(matched)}")
print(f"Images without labels: {len(unmatched_images)}")
print(f"Labels without images: {len(unmatched_labels)}")

if unmatched_images:
    print(f"\nFirst 5 images without labels:")
    print(list(unmatched_images)[:5])

if unmatched_labels:
    print(f"\nFirst 5 labels without images:")
    print(list(unmatched_labels)[:5])
```
